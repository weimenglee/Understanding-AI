Use a web browser and go to: https://teachablemachine.withgoogle.com

Click **Get Started**

Click **Image Project**

Click **Standard Image model**

- Click Class 1 and name it as **Barack Obama**
- Click Class 2 and name it as **Donald Trump**
- Click Add a class
- Click Class 3 and name it as **Tom Cruise**

 - Your project should now look like the following.
 - Click the Upload button for **Barack Obama** and drag and drop the images for Barack Obama
 - Click the Upload button for **Donald Trump** and drag and drop the images for Donald Trump
 - Click the Upload button for **Tom Cruise** and drag and drop the images for Tom Cruise    

Your project should now look like the following:

Click the **Train Model** button

Once the model is trained, click the **Webcam** dropdown:

Select **File**

- Drag and image and drop it onto the box titled **"Choose images from your files, or drag and drop here"**.
- You should now see the prediction
