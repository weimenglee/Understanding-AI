# Plotting the Data

```python
import numpy as np
import matplotlib.pyplot as plt

# Create simple dataset
# Hours studied (X) vs Exam scores (y)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # independent variable
y = np.array([20, 45, 66, 88, 100])           # dependent variable

# Visualize results
plt.scatter(X, y, color="blue", label="Actual data")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.legend()
plt.show()
```

 # Simple Linear Regression

```python
from sklearn.linear_model import LinearRegression

# Train the linear regression model
model = LinearRegression()
model.fit(X, y)

# Make predictions
predictions = model.predict(X)

# Visualize results
plt.scatter(X, y, color="blue", label="Actual data")
plt.plot(X, predictions, color="red", label="Regression line")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.legend()
plt.show()

# Step 6: Try predicting a new value
new_hours = np.array([[4.5]])
predicted_score = model.predict(new_hours)
print(f"Predicted score for 6 hours of study: {predicted_score[0]:.2f}")
```

# Polynomial Regression

```python
from sklearn.preprocessing import PolynomialFeatures

# Transform features into polynomial terms
poly = PolynomialFeatures(degree=2)   # quadratic curve
X_poly = poly.fit_transform(X)

# Train polynomial regression model
model = LinearRegression()
model.fit(X_poly, y)

# Make predictions
X_range = np.linspace(1, 5, 100).reshape(-1, 1)
X_range_poly = poly.transform(X_range)
predictions = model.predict(X_range_poly)

# Visualize results
plt.scatter(X, y, color="blue", label="Actual data")
plt.plot(X_range, predictions, color="red", label="Polynomial regression curve")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.legend()
plt.show()

# Try predicting a new value
new_hours = np.array([[4.5]])
new_hours_poly = poly.transform(new_hours)
predicted_score = model.predict(new_hours_poly)
print(f"Predicted score for 4.5 hours of study: {predicted_score[0]:.2f}")
```
