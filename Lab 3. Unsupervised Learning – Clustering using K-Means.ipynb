# Generating Blobs

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs

blobs, classes = make_blobs(500, centers=4)  # Generate isotropic Gaussian 
                                             # blobs for clustering
print(blobs)

# the Classes shows which group each point falls into
print(classes)
```

# Plotting the Blobs

```python
f, ax = plt.subplots(figsize=(7.5, 7.5))

# rgb = np.array(['r', 'g', 'b', 'm'])
rgb = np.array(['r', 'r', 'r', 'r'])

#plot the blobs using a scatter plot and use color coding
ax.scatter(blobs[:, 0], blobs[:, 1], color=rgb[classes])
ax.set_title("Blobs")
```

# Finding and Displaying the Center Points for the Blobs

```python
from sklearn.cluster import KMeans
import pandas as pd

# use K-Means and assume a cluster size of 3
kmean = KMeans(n_clusters=3, n_init='auto')
kmean.fit(blobs)

f, ax = plt.subplots(figsize=(7.5, 7.5))
ax.scatter(blobs[:, 0], blobs[:, 1], color=rgb[classes])

# plot the 3 center points
ax.scatter(kmean.cluster_centers_[:, 0], 
           kmean.cluster_centers_[:, 1], 
           marker='*', 
           s=250, 
           color='black', 
           label='Centers')

ax.set_title("Blobs")
ax.legend(loc='best')

print(kmean.cluster_centers_)  # print out the 3 center points
```
