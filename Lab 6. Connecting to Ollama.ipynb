# Maintaining Conversation

```python
import ollama

messages = [
    {
        "role": "system", 
        "content": "You are a sarcastic AI assistant"
    }]   

while True:
    prompt = input('\nAsk a question: ') 
    if prompt == "quit":
        break
        
    messages.append(
        {
            'role':'user',
            'content':prompt
        })
    
    response = ollama.chat(
        model = "llama3.2", 
        messages = messages
    )    
    print(response["message"]["content"])    
    messages.append(
        {
            'role':'assistant',
            'content':response["message"]["content"]
        })
```

# Streaming

```python
import ollama

# Simple completion
response = ollama.chat(model = "llama3.2", 
   messages=[
       {
           "role": "system", 
           "content": "You are a sarcastic AI assistant"
       },
       {
           "role": "user", 
           "content": "Which is better? Mac or Windows"
       }
   ],
  stream = True)

for chunk in response:
    # Each chunk contains a partial message
    if "message" in chunk and "content" in chunk["message"]:
        print(chunk["message"]["content"], end="", flush=True)

print()  # Newline after stream completes
```
